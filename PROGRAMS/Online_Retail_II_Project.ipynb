```python
import pandas as pd

file_path = "online_retail_ii.xlsx"

# Display the sheet names
excel_file = pd.ExcelFile(file_path)
print("Sheet names:", excel_file.sheet_names)

# Load each sheet
data = pd.read_excel(file_path, sheet_name=None)

# Check rows and columns
for sheet_name, df in data.items():
    print("\nSheet:", sheet_name)
    print("Rows and columns:", df.shape)
    display(df.head())
```

```python
# Combine the two sheets into one DataFrame
df = pd.concat(data.values(), ignore_index=True)

# Check the combined dataset
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

```python
# Check column names
print("Columns:")
print(df.columns.tolist())

# Check data types
print("\nData types:")
print(df.dtypes)

# Count missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Summary statistics
print("\nSummary statistics:")
display(df.describe(include="all").T)
```

```python

# 1. Summary statistics for numeric columns
print("Numeric summary:")
display(df[["Quantity", "Price", "Customer ID"]].describe().T)

# 2. Check negative and zero quantities
print("\nNegative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())

# 3. Check negative and zero prices
print("\nNegative prices:", (df["Price"] < 0).sum())
print("Zero prices:", (df["Price"] == 0).sum())

# 4. Check cancellation invoices
print("\nInvoices starting with C:")
print(df["Invoice"].astype(str).str.startswith("C").sum())

# 5. Display some cancellation records
cancellations = df[
    df["Invoice"].astype(str).str.startswith("C")
]

display(cancellations.head(10))

# 6. Display records with missing descriptions
print("\nRows with missing descriptions:")
display(df[df["Description"].isnull()].head(10))
```

```python

# Work on a copy so the original data stays unchanged
check_df = df.copy()

# Create diagnostic flags
check_df["Invoice"] = check_df["Invoice"].astype(str)

check_df["IsCancellationInvoice"] = (
    check_df["Invoice"].str.startswith("C")
)

check_df["IsNegativeQuantity"] = (
    check_df["Quantity"] < 0
)

check_df["IsZeroPrice"] = (
    check_df["Price"] == 0
)

check_df["IsNegativePrice"] = (
    check_df["Price"] < 0
)

# Count each category
print("Cancellation invoice flag:")
print(check_df["IsCancellationInvoice"].value_counts())

print("\nNegative quantity flag:")
print(check_df["IsNegativeQuantity"].value_counts())

print("\nZero price flag:")
print(check_df["IsZeroPrice"].value_counts())

print("\nNegative price flag:")
print(check_df["IsNegativePrice"].value_counts())
```

```python

# Compare cancellation invoices and negative quantities
print("Cancellation invoice AND negative quantity:")
print(
    (
        check_df["IsCancellationInvoice"]
        & check_df["IsNegativeQuantity"]
    ).sum()
)

print("\nCancellation invoice but NOT negative quantity:")
print(
    (
        check_df["IsCancellationInvoice"]
        & ~check_df["IsNegativeQuantity"]
    ).sum()
)

print("\nNegative quantity but NOT cancellation invoice:")
print(
    (
        ~check_df["IsCancellationInvoice"]
        & check_df["IsNegativeQuantity"]
    ).sum()
)

# Show all records with negative prices
print("\nRecords with negative prices:")
display(check_df[check_df["IsNegativePrice"]])
```

```python

# Inspect the 10 largest quantities
print("Largest quantities:")
display(
    df.sort_values("Quantity", ascending=False).head(10)
)

# Inspect the 10 smallest quantities
print("\nSmallest quantities:")
display(
    df.sort_values("Quantity").head(10)
)

# Inspect the 10 largest prices
print("\nLargest prices:")
display(
    df.sort_values("Price", ascending=False).head(10)
)

# Inspect the 10 smallest prices
print("\nSmallest prices:")
display(
    df.sort_values("Price").head(10)
)
```

```python
# Make a working copy of the original dataset
sales_df = df.copy()
sales_df["Invoice"] = sales_df["Invoice"].astype(str)
sales_df = sales_df[
    (~sales_df["Invoice"].str.startswith("C"))
    & (sales_df["Quantity"] > 0)
    & (sales_df["Price"] > 0)
    & (sales_df["Description"].notna())
].copy()

# Calculate line revenue
sales_df["Revenue"] = (
    sales_df["Quantity"] * sales_df["Price"]
)

# Display the resulting dataset
print("Original rows:", len(df))
print("Positive-sales rows:", len(sales_df))
print("Rows excluded from this subset:", len(df) - len(sales_df))

print("\nFirst 5 rows:")
display(sales_df.head())

print("\nRevenue summary:")
display(sales_df["Revenue"].describe())
```

```python

# Check whether the selected rows meet our criteria
print("Cancellation invoices remaining:",
      sales_df["Invoice"].str.startswith("C").sum())

print("Non-positive quantities remaining:",
      (sales_df["Quantity"] <= 0).sum())

print("Non-positive prices remaining:",
      (sales_df["Price"] <= 0).sum())

print("Missing descriptions remaining:",
      sales_df["Description"].isna().sum())

# Export this analysis subset to CSV
sales_df.to_csv(
    "online_retail_sales_clean.csv",
    index=False
)

# Check exported file size
from pathlib import Path

output_file = Path("online_retail_sales_clean.csv")

print("\nExported file:", output_file.name)
print("File size (MB):",
      round(output_file.stat().st_size / (1024 * 1024), 2))
```

```python

# Count exact duplicate rows in the sales subset
duplicate_count = sales_df.duplicated().sum()

print("Exact duplicate rows in sales subset:",
      duplicate_count)

# Display some examples without deleting them
print("\nExample duplicate rows:")

duplicate_rows = sales_df[
    sales_df.duplicated(keep=False)
]

display(duplicate_rows.head(20))
```

```python

# Count repeated invoice and stock-code combinations
invoice_product_counts = (
    sales_df.groupby(["Invoice", "StockCode"])
    .size()
    .reset_index(name="Count")
)

repeated_invoice_products = invoice_product_counts[
    invoice_product_counts["Count"] > 1
]

print("Repeated invoice and product combinations:",
      len(repeated_invoice_products))

display(repeated_invoice_products.head(20))
```

```python

# Use the original 8 transaction columns to identify duplicates
original_columns = [
    "Invoice",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "Price",
    "Customer ID",
    "Country"
]

# Count exact duplicates using the original columns
duplicate_count = sales_df.duplicated(
    subset=original_columns
).sum()

print("Exact duplicate rows:", duplicate_count)

# Create a separate deduplicated version
sales_dedup_df = sales_df.drop_duplicates(
    subset=original_columns,
    keep="first"
).copy()

print("\nRows before deduplication:", len(sales_df))
print("Rows after deduplication:", len(sales_dedup_df))
print("Rows removed:", len(sales_df) - len(sales_dedup_df))
```

```python

# Calculate total revenue before and after deduplication
revenue_before = sales_df["Revenue"].sum()
revenue_after = sales_dedup_df["Revenue"].sum()

revenue_difference = revenue_before - revenue_after

print("Revenue before deduplication:",
      round(revenue_before, 2))

print("Revenue after deduplication:",
      round(revenue_after, 2))

print("Revenue difference:",
      round(revenue_difference, 2))

if revenue_before != 0:
    percentage_difference = (
        revenue_difference / revenue_before
    ) * 100

    print("Percentage difference:",
          round(percentage_difference, 2), "%")
```

```python

# Count occurrences of each original transaction record
duplicate_groups = (
    sales_df.groupby(original_columns, dropna=False)
    .size()
    .reset_index(name="Occurrences")
)

# Keep records that occur more than once
duplicate_groups = duplicate_groups[
    duplicate_groups["Occurrences"] > 1
]

print("Number of repeated transaction groups:",
      len(duplicate_groups))

print("\nOccurrences per repeated group:")
display(duplicate_groups["Occurrences"].value_counts().sort_index())

print("\nExamples of repeated groups:")
display(duplicate_groups.head(10))
```

```python

# Create a copy for monthly analysis
monthly_df = sales_df.copy()

# Extract year and month from invoice date
monthly_df["YearMonth"] = (
    monthly_df["InvoiceDate"].dt.to_period("M").astype(str)
)

# Calculate monthly revenue
monthly_revenue = (
    monthly_df.groupby("YearMonth")["Revenue"]
    .sum()
    .reset_index()
)

print("Monthly revenue:")
display(monthly_revenue)

# Check the month with the highest revenue
highest_month = monthly_revenue.loc[
    monthly_revenue["Revenue"].idxmax()
]

print("\nMonth with highest revenue:")
print(highest_month)
```
